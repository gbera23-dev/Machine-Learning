### Load Best Model's Pipeline

```python

import mlflow.sklearn

model_id_uri = "models:/m-ab6947cd1d714618ae645f3e79797464"

loaded_model = mlflow.sklearn.load_model(model_id_uri)
```

```python
df_transaction = pd.read_csv('/kaggle/input/competitions/ieee-fraud-detection/test_transaction.csv')
df_identity = pd.read_csv('/kaggle/input/competitions/ieee-fraud-detection/test_identity.csv')

df = df_transaction.merge(df_identity, on='TransactionID', how='left')

df = df.rename(columns=lambda x: x.replace('-', '_') if x.startswith('id') else x)


preds = loaded_model.predict_proba(df)

submission = pd.DataFrame({
    'TransactionID': df['TransactionID'],
    'isFraud': preds[:, 1]
})

submission.to_csv('submission.csv', index=False)
print("Submission file created successfully!")
```
